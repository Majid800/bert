```python
from transformers import pipeline

print("Loading local BioBERT model...")

ner_pipeline = pipeline(
    task="token-classification",
    model="d4data/biomedical-ner-all",
    aggregation_strategy="simple",
)

print("Model loaded and ready for local inference.")
```

```python
def inspect_extraction(input_text):
    print(f"Input: {input_text}\n")
    
    # 1. Run raw pipeline inference
    raw_entities = ner_pipeline(input_text)

    if not raw_entities:
        print("  No entities detected.\n")
        return

    # 2. Force strict left-to-right reading order using start character indices
    ordered_entities = sorted(raw_entities, key=lambda x: x["start"])

    # 3. Print entities in exact sentence order
    for item in ordered_entities:
        span = item["word"]
        label = item["entity_group"]
        confidence = item["score"]
        start_pos = item["start"]
        end_pos = item["end"]
        
        # Display with character offsets [start:end] to verify sequential alignment
        print(
            f"  [{start_pos:>3}:{end_pos:<3}] "
            f"Extracted: {span:<32} "
            f"Label: {label:<22} "
            f"Confidence: {confidence:.2f}"
        )
    print("\n")
```

```python
inspect_extraction("Hi, I am 19. My asthma has been terrible lately. I'm using my Fostair 100/6 inhaler 2 puffs twice a day every day, and I also use it when I feel short of breath. I haven't taken any other tablets or add-on inhalers. I have symptoms literally every single day of the week, and I wake up breathless probably 3 or 4 nights a week. I can't even walk to uni or do basic household chores without stopping to catch my breath, and exercise is completely out of the question right now. Smoke and dust always set me off. I had a really bad turn last month where I had to take a 5-day course of prednisolone from the GP, but I didn't go to hospital or urgent care. No severe attack right now, just chronically bad.")
```
